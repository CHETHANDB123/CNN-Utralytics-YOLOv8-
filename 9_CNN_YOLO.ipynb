##Load Model

```python
from ultralytics import YOLO
model=YOLO("yolov8n.pt")
```

##Detection

```python
results=model("/content/insert-8.webp")
# Display image manually (like  YOLOv4)
import matplotlib.pyplot as plt
img=results[0].plot()
plt.figure(figsize=(10,6))
plt.imshow(img)
plt.axis('off')
plt.show()
```
